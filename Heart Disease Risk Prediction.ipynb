# Understanding the features

1.   age: the person's age in years.
2.   sex: the person's sex (1= male,0=female)
3.   cp: the chest pain experineced (Value 1: typical angina, Value 2: atypical angina, Value 3: non-anginal pain, Value 4: asymptomatic)
4.   trestbps: the person's resting blood pressure (in mm Hg)
5.   chol: the person's cholesterol measurement (in mg/dl)
6.   fbs: the person's fasting blood sugar > 120 mg/dl (1 = true, 0 = Flase)
7.   restecg: resting electrocardiographic measurement ( Value 0: Normal, Value 1: Having ST-T wave abnormality, Value 2: showing probable or definite left ventricular hypertropy by Estes' criteria)
8.   thalach: the person's maximum heart achieved (bpm)
9.   exang: exercise induced angina ( 1 = yes, 0 = no)
10.  oldpeak: ST depression induced by exercise relative to rest ('ST' relates to position on ECG plot.)
11.  slope: the slope of the peak exercise ST segment ( Value 1: Unsloping, Value 2: Flat, Value 3: Down-Sloping)
12.  ca: number of major vessels (0-3) colored by fluoroscopy
13.  thal: a blood disorder called thalassemia ( 3 = normal, 6 = fixed defect, 7 = reversible defect)
14.  target: heart disease ( 0=no,1=yes)

### Splitting the features and target

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
scaler= StandardScaler()
```

```python
X = df.drop(columns = 'target',axis=1)
Y = df['target']
```

```python
X1= scaler.fit_transform(X)
```

```python
X = X1
```

```python
X_train,X_test,Y_train,Y_test =train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
## Model training
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
logregressor = LogisticRegression()
```

```python
# training the logistic regression model with training data
```

```python
logregressor.fit(X_train,Y_train)
```

```python
## Model evaluation
```

```python
X_train_prediction = logregressor.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction,Y_train)
```

```python
print(training_data_accuracy)
```

```python
X_test_prediction = logregressor.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction,Y_test)
```

```python
print(test_data_accuracy)
```
